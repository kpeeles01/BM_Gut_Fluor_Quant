```python
#When there are 4 channels run this script for Cy3 fluorescence
#When there are 3 channels run this script for endogenous GFP fluorescence

from pathlib import Path
import pandas as pd

def int_per_pxl(LocalDir, PredClass):
    '''
    This function takes the following arguments:
        LocalDir: the directory where all csv files from ilastik object detection are stored. 
                    other file formats can also exist in the directory
        PredClass: wanted category from ilastik. case sensitive. 
    and returns:
        an excel sheet with file name, total size in pixels total intensity in the green channle and intensity per pixel.
    '''
    
    local_dir = LocalDir
    csv_files = Path(local_dir).glob("*.csv")

    total_size_in_pxl = []
    total_green_intensity = []
    csv_file_names = []

    for csv in csv_files:
        csv = str(csv) 
        df = pd.read_csv(csv)
        wanted = df.loc[df["Predicted Class"] == PredClass]
        if len(wanted.index) == 0:
            print("None of the values in {} falls into this category.".format(csv.split("/")[-1]))
        else:
            total_size_in_pxl.append(wanted["Size in pixels"].sum())
            total_green_intensity.append(wanted["Total Intensity_1"].sum())
            intensity_per_pxl = [i / j for i, j in zip(total_green_intensity, total_size_in_pxl)]
            csv_file_names.append(csv.split("\\")[-1])

    result = {
        "file name": csv_file_names,
        "total size in pixels": total_size_in_pxl,
        "total intensity in channel 1": total_green_intensity,
        "intensity per pixel": intensity_per_pxl
    }
    result_df = pd.DataFrame(result)
    result_df.set_index("file name", inplace=True)
    
    return result_df.to_excel(local_dir + "/GFP Fluor CSper vkgGFP DD R1.xlsx")
    #change the file here ^^ to change the name of the file
```
